# Classroom Workshop
This workshop uses a synthetic e-commerce dataset modelled after a Southeast Asian online retail platform. It is designed to teach star schema analytics — a data warehouse pattern where a central fact table (transactional events) is surrounded by dimension tables (descriptive context). All queries execute in Databricks SQL against Delta Lake tables in Unity Catalog.

## Data Ingestion

Follow these steps in Databricks Free Edition before running any SQL cells.



### Create Schema in Databricks

Step 1 — Create the Schema (ecommerce)
- Click the Catalogue icon on the left navigation bar. 
- Click the Create Schema button (or ➕ icon next to the workspace catalog).
- Enter schema name: ecommerce. Leave Storage Location as default. Click Create.



### Create Managed Tables in Databricks
Once you have created the schema ecommrce, select the schema and open the schema window. Here we will create four managed tables as listed below:
- customers
- order_items
- orders
- products


Step 2 - Create four tables
- Click the Create Table button (or ⁞ icon next to the schema screen and choose create teable in the menu item).
- Drag the customers.csv file from the workshop/data folder and drop in the file selection window. 
- Enter table name: customers. Accept the default columns, datatype and description. Confirm by clicking Create.

Step 3: Repeat the same steps above and create the other three tables mentioned.

## Verify Data

Connection check: Run this quick JOIN validation to confirm orders and customers connect correctly: SELECT COUNT(*) FROM orders o JOIN customers c ON o.customer_id = c.customer_id;  Expected result: 200 rows — matching all 200 orders to a customer.

```python
%sql
SELECT 'customers'   AS tbl, COUNT(*) AS rows FROM ecommerce.customers
UNION ALL
SELECT 'products',          COUNT(*)         FROM ecommerce.products
UNION ALL
SELECT 'orders',            COUNT(*)         FROM ecommerce.orders
UNION ALL
SELECT 'order_items',       COUNT(*)         FROM ecommerce.order_items
ORDER BY rows DESC;
```
