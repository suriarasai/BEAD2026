```sql
-- Select all columns from customers table
SELECT * 
FROM workspace.ecommerce.customers;

-- Select specific columns
SELECT customer_id, first_name, last_name, email 
FROM workspace.ecommerce.customers;

-- Filter customers by country
SELECT * 
FROM workspace.ecommerce.customers
WHERE country = 'Singapore';

-- Count customers by membership tier
SELECT membership_tier, COUNT(*) AS customer_count
FROM workspace.ecommerce.customers
GROUP BY membership_tier;

-- Find customers registered after Jan 1, 2026
SELECT customer_id, registration_date
FROM workspace.ecommerce.customers
WHERE registration_date > '2020-01-01';
```
